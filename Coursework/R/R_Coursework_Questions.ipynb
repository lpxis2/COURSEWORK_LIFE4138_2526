### Question 6

Using the `starwars` dataset, identify the individuals with:

- the tallest height
- the lowest mass

HINT: the starwars dataset is part of the dplyr package

```r
# loading dplyr to get the starwars dataset
library(dplyr)
data(starwars)

# finding the biggest value in the height column of the dataset and ignoring NA values
max_height <- max(starwars$height, na.rm = TRUE)

# filtering to the row with the tallest height
tallest_individual <- starwars %>%
  filter(!is.na(height)) %>%   # removes NA
  filter(height == max_height) %>%   # keeping the tallest
  select(name, height)  

# printing tallest individual with message to make it clear
cat("Tallest individual:\n")
print(tallest_individual)

# working out smallest value in mass column and ignoring NA values
min_mass <- min(starwars$mass, na.rm = TRUE)

# filtering to the row with the lowest mass
lightest_individual <- starwars %>%
  filter(!is.na(mass)) %>%                 # removes NA values
  filter(mass == min_mass) %>%             # keeps lightest
  select(name, mass)   

# printing lightest individual with message to make it clear
cat("Lightest individual:\n")
print(lightest_individual)
```

### Question 7

The `dplyr` dataset `storms` is part of the US National Oceanic and Atmospheric Administration database of Atlantic hurricanes. It contains multiple repeated measurements from hurricanes. For a given storm, there are measurements from every 6 hours across the life time of the storm.

Use a tidyverse approach with `dplyr` to answer the following:

- how many different storms are contained in the database?
- which year had the most storms?
- which storm, in which year had the highest pressure?

HINT:
unique() will help ensure you remove duplicated data from a vector e.g.
x <- c(1, 1, 2, 2, 3, 3, 4, 4)
unique(x)

```r
# loading dplyr to get storms dataset
library(dplyr)
data(storms)

# because storm has repeated measurements every 6 hours, need to remove duplicates so only unique names stay
n_storms <- length(unique(storms$name))
# adding message to make it clearer
cat("Number of different storms:", n_storms, "\n")

# making a unique identifier for each storm in each year. sep will put an underscore btwn them to make it easier to just get year later
storm_id <- paste(storms$year, storms$name, sep = "_")
# unique will remove duplicates so each storm/year combo appears once
storm_id_unique <- unique(storm_id)
# putting unique ids into a df 
storm_id_df <- data.frame(storm_id = storm_id_unique)
# need to just get the year part of the id so going to use sub_.* 
# everything is text right now so have to turn it numeric
storm_id_df$year <- as.numeric(sub("_.*", "", storm_id_df$storm_id))
# counting the number of unique storms per year and sorting from biggest to smallest value
storms_per_year <- storm_id_df %>%
  count(year) %>%          # creating column called n = how many in each year
  arrange(desc(n))         
# printing with message
cat("year wirh most storms:", storms_per_year$year[1], "\n")
```

```r
library(dplyr)
data(storms)

# working out highest value in pressure column
max_pressure <- max(storms$pressure, na.rm = TRUE)

# finding a storm and year that matches max value
highest_pressure_storm <- storms %>%
  filter(!is.na(pressure)) %>%          # removes missing values
  filter(pressure == max_pressure) %>%  # keep only the max pressure rows
  select(year, name, pressure)          
# printing 
highest_pressure_storm
```

The `ggplot2` dataset `diamonds` contains the price and information from the sale of 54,000 different diamonds. You can use ?diamonds to find out more about the dataset if you want.

Using `ggplot2`:

- plot the relationship between the carat (i.e. the weight of the diamond) and the price.
- plot a histogram of prices of all the diamonds
- plot a boxplot of the difference in price between the different cuts

```r
# loading ggplot2 for diamonds data and plotting tools
library(ggplot2)
data(diamonds)

# making a scatterplot 
ggplot(aes(x = carat, y = price), data = diamonds) +
  geom_point()

# making a histogram of prices 
ggplot(aes(x = price), data = diamonds) +
  geom_histogram()

# making bocplot of price differences across cuts
ggplot(aes(x = cut, y = price), data = diamonds) +
  geom_boxplot()
```

### Question 9

Identify the maximum and minimum price that the diamonds in the diamonds data set sold for. Add a new column to the diamonds data set that contains the string “Low price” if the value in the price column is less than 5000 dollars and “High price” if it is above 5000 dollars.

```r
# loading ggplot2 for plotting tools and diamonds dataset
library(ggplot2)
data(diamonds)

# finding min and max values
min_price <- min(diamonds$price, na.rm = TRUE) 
max_price <- max(diamonds$price, na.rm = TRUE) 

# printing with message
cat("Min price:", min_price, "\n")
cat("Max price:", max_price, "\n")

# adding a new column using ifelse 
diamonds$price_class <- ifelse(diamonds$price < 5000, "Low price", "High price")

head(diamonds) # lets us look at dataset to make sure new column was added
```

### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
# loading ggplot2 for plotting tools and diamonds dataset
library(ggplot2)
data(diamonds)

# writing function to convert carat to grams
carat_to_grams <- function(carat){
  grams <- carat * 0.2
  return(grams)
}

# creating new columns using the function
diamonds$weight_g <- carat_to_grams(diamonds$carat)

# heaviest diamond, finding max weight 
max_weight_g <- max(diamonds$weight_g, na.rm = TRUE)
# printing output with message
cat("Heaviest diamond (grams):", max_weight_g, "\n")
```
