### Question 6
Using the startwars dataset imported below, identify the individuals with tallest height and lowest mass.

````python
starwars_dataset="https://www.fabricionarcizo.com/post/starwars/updated_starwars.csv"
starwars = pd.read_csv(starwars_dataset)
````

````python
# individuals with tallest height 

max_height = starwars['height'].max() #finds maximum value
tallest_individuals = starwars[starwars['height'] == max_height] # filters rows matching max height
tallest_individuals.head ()
````

````python
# individuals with lowest mass

min_mass = starwars['mass'].min() #finds minimum mass
lowest_mass_individuals = starwars[starwars['mass'] == min_mass] #filters rows matching min mass
lowest_mass_individuals.head ()
````

### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

- Calculate the number of unique countries in the dataset.

- Calculate the country and year with the highest life expectancy.

- Calculate the year with the highest total expenditure.

- Provide your code using pandas.


````python
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')
````

````python
# number of unique countries

unique_countries = life_expectancy['Country'].nunique()
print("Number of Unique Countries:", unique_countries)

# country and year with highest life expectancy

max_life_exp = life_expectancy ['Life_Expectancy'].max()
highest_life_exp_row = life_expectancy[life_expectancy['Life_Expectancy'] == max_life_exp]
print("Country and year with highest life expectancy:",
      highest_life_exp_row["Country"].iloc[0],
      highest_life_exp_row["Year"].iloc[0])

# year with highest total expenditure
highest_total_expenditure = life_expectancy.groupby('Year')["Spending_USD"].sum()
year_highest_spend = highest_total_expenditure.idxmax()
print("Year with highest total expenditure:", year_highest_spend)
````

### Question 8
Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

1. Carat vs Price in a scatter plot.
2. A histogram of prices.
3. A boxplot of cut vs price.

The code can be loaded from seaborn in the cell below.

````python
diamonds = sns.load_dataset('diamonds')
````

````python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

# scatter plot of carat vs price 
plt.scatter(diamonds['carat'], diamonds ['price'], color='orange')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.title('Carat vs Price')
plt.show()

#histogram

diamonds['price'].plot.hist(bins=20, edgecolor='black')
plt.title('Distribution of Diamond Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()

# boxplot
diamonds.boxplot(column= 'price', by='cut', grid=False)
plt.title('Price vs Cut Quality')
plt.xlabel('Cut')
plt.ylabel('Price')
plt.show()
````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
def carat_gram (carat):
    return carat * 0.2

diamonds['weight_g'] = diamonds ['carat'].apply(carat_gram)

heaviest_diamond = diamonds.loc[diamonds['weight_g'].idxmax()]

print("Heaviest diamond (carat, grams):", heaviest_diamond["carat"], heaviest_diamond["weight_g"])
````
